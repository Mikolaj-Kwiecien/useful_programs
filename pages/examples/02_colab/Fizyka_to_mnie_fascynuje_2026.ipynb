## Trochę matematyki

### Pochodne

Pochodna funkcji $f(x)$ to funkcja $f'(x)$, która opisuje tempo zmian wartości funkcji $f(x)$ w zależności od zmiany argumentu $x$.

$$
\begin{array}{|c|c|}
\hline
\textbf{Funkcja } f(x) & \textbf{Pochodna } f'(x) \\ \hline
x^n & n \cdot x^{n-1} \\ \hline
\sin(x) & \cos(x) \\ \hline
\cos(x) & -\sin(x) \\ \hline
e^x & e^x \\ \hline
\ln(x) & \frac{1}{x} \\ \hline
\frac{1}{x} & -\frac{1}{x^2} \\ \hline
a^x & a^x \ln(a) \\ \hline
\tan(x) & \frac{1}{\cos^2(x)} \\ \hline
\arcsin(x) & \frac{1}{\sqrt{1 - x^2}} \\ \hline
\arccos(x) & -\frac{1}{\sqrt{1 - x^2}} \\ \hline
\arctan(x) & \frac{1}{1 + x^2} \\ \hline
\end{array}
$$

Porównanie:

$$
\begin{array}{|c|c|c|}
\hline
\textbf{Kategoria} & \textbf{Matematycy} & \textbf{Fizycy} \\ \hline
\textbf{Zmienne} & x & t \\ \hline
\textbf{Funkcje} & y(x) & x(t) \\ \hline
\textbf{Pochodna pierwsza} & \frac{dy}{dx} & \frac{dx}{dt} \quad \text{(prędkość)} \\ \hline
\textbf{Pochodna druga} & \frac{d^2y}{dx^2} & \frac{d^2x}{dt^2} \quad \text{(przyspieszenie)} \\ \hline
\textbf{Zastosowanie} & \text{Analiza matematyczna} & \text{Dynamika ciał} \\ \hline
\end{array}
$$

Naturalnie ma wiele zastosowań!


### Równania algebraiczne

Równania ze zmienną $x$ to podstawowy rodzaj równań, w których rozwiązanie polega na znalezieniu liczby, która zapewnia tożsamościowe spełnienie równania – oznacza to, że lewa strona równa się prawej stronie dla danej wartości $x$.

Przykład równania algebraicznego:

$$
2x + 3 = 7
$$

Rozwiązaniem jest $x = 2$, ponieważ po podstawieniu tej wartości obie strony równania są równe:

$$
2 \cdot 2 + 3 = 7.
$$

### Równania różniczkowe

Równania różniczkowe to bardziej zaawansowany typ równań, w których zamiast szukać liczby, szukamy funkcji $x(t)$, która spełnia pewną relację różniczkową. Rozwiązanie takiego równania polega na znalezieniu funkcji, która zapewnia tożsamościowe spełnienie równania dla każdej wartości zmiennej niezależnej $t$.

Przykład równania różniczkowego:

$$
\frac{d^2 x(t)}{dt^2} = - x(t) - 2 \frac{dx(t)}{dt}.
$$

Tutaj szukamy funkcji $x(t)$, która spełnia to równanie – np. funkcja wykładnicza $x(t) = e^{-t}$ może być rozwiązaniem dla odpowiednich warunków początkowych.

Równania różniczkowe mają zastosowanie w opisie wielu zjawisk fizycznych, takich jak ruch, przenoszenie ciepła, czy dynamika populacji, i stanowią podstawę matematycznego modelowania w naukach przyrodniczych i technicznych.


## Trochę fizyki — ruch i to, co na niego wpływa

**Ruch** jest to zmiana położenia ciała w czasie. Położenie ciała określamy względem ustalonego układu odniesienia (ruch jest zawsze względem czegoś).

Prędkość to zmiana położenia w czasie. Przyspieszenie to zmiana prędkości w czasie.

### Historia

- **Arystoteles**: Dużo rozmyślań na temat ruchu.
- **Galileusz**: Liczne obserwacje i eksperymenty dotyczące ruchu.
- **Newton**: Ujęcie koncepcji ruchu w sposób matematyczny.

### Drugie prawo Newtona

Przyspieszenie obiektu jest wprost proporcjonalne do siły netto działającej na niego i odwrotnie proporcjonalne do jego masy:

$$
ma = \text{Siła}
$$

### Isaac Newton

*"Podajcie mi wszystkie siły działające na ciało, a przewidzę z boską/matematyczną precyzją, jak ciało się poruszy"*

$$
m \frac{d^2 x(t)}{dt^2} = \text{Siła}
$$

Rozwiązanie równania różniczkowego dla zadanej siły daje:

$$
x(t) = \dots \text{(funkcja zależna od czasu)}
$$

(plus warunki początkowe... o tym później)

**Newton dostrzegł, jak działa tkanka Wszechświata** na fundamentalnym poziomie, jeśli chodzi o dynamikę ciał!


### Ruch w jednym wymiarze

Położenie to wszystko, czego potrzebujemy, aby opisać ruch w jednym wymiarze. Możemy je opisać jako funkcję położenia $x(t)$, prędkości $v(t)$ i przyspieszenia $a(t)$.

$$
\begin{aligned}
x(t) & & & \text{Położenie} \\
v(t) & = \frac{dx(t)}{dt} &  &\text{Prędkość} \\
a(t) & = \frac{dv(t)}{dt} = \frac{d^2 x(t)}{dt^2}  & & \text{Przyspieszenie}
\end{aligned}
$$

Tabela z paroma przykładami:

$$
\begin{array}{|c|c|c|}
\hline
\textbf{Położenie } x(t) & \textbf{Prędkość } v(t) & \textbf{Przyspieszenie } a(t) \\ \hline
x_0 \text{ (stała)} & 0 & 0 \\ \hline
x_0 + vt & v \text{ (stała)} & 0 \\ \hline
x_0 + v_0t + \frac{1}{2}at^2 & v_0 + at & a \text{ (stała)} \\ \hline
A \sin(\omega t + \phi) & A \omega \cos(\omega t + \phi) & -A \omega^2 \sin(\omega t + \phi) \\ \hline
e^{kt} & ke^{kt} & k^2 e^{kt} \\ \hline
\end{array}
$$


### Podstawy ruchu

- **Prędkość** wpływa na **położenie**.
- **Siła** wpływa na **prędkość**.

### Rzut ukośny bez oporu i z oporem powietrza

**Polecenie dla ChatGPT:**

> Podaj kod Pythona, który pokazuje rzut ukośny bez oporu i z oporem powietrza.


```python
import numpy as np
import matplotlib.pyplot as plt

# Parametry rzutu
v0 = 50  # prędkość początkowa (m/s)
angle = 45  # kąt rzutu (stopnie)
g = 9.81  # przyspieszenie grawitacyjne (m/s^2)
mass = 0.5  # masa obiektu (kg)
drag_coefficient = 0.1  # współczynnik oporu powietrza (kg/m)

# Funkcje pomocnicze
def trajectory_no_air_resistance(v0, angle, g, t):
    angle_rad = np.radians(angle)
    x = v0 * np.cos(angle_rad) * t
    y = v0 * np.sin(angle_rad) * t - 0.5 * g * t**2
    return x, y

def trajectory_with_air_resistance(v0, angle, g, mass, drag_coefficient, dt=0.01):
    angle_rad = np.radians(angle)
    vx = v0 * np.cos(angle_rad)
    vy = v0 * np.sin(angle_rad)
    x, y = 0, 0
    x_positions, y_positions = [x], [y]

    while y >= 0:
        ax = -drag_coefficient * vx / mass
        ay = -g - (drag_coefficient * vy / mass)
        vx += ax * dt
        vy += ay * dt
        x += vx * dt
        y += vy * dt
        x_positions.append(x)
        y_positions.append(y)

    return x_positions, y_positions

# Czas symulacji dla przypadku bez oporu powietrza
t_max = 2 * v0 * np.sin(np.radians(angle)) / g
t = np.linspace(0, t_max, num=500)

# Obliczenia
x_no_air, y_no_air = trajectory_no_air_resistance(v0, angle, g, t)
x_with_air, y_with_air = trajectory_with_air_resistance(v0, angle, g, mass, drag_coefficient)

# Wykres
plt.figure(figsize=(10, 6))
plt.plot(x_no_air, y_no_air, label="Bez oporu powietrza", linestyle="--")
plt.plot(x_with_air, y_with_air, label="Z oporem powietrza", linestyle="-")
plt.title("Rzut ukośny: porównanie z i bez oporu powietrza")
plt.xlabel("Odległość (m)")
plt.ylabel("Wysokość (m)")
plt.legend()
plt.grid()
plt.show()
```

### Ruch ciała na sprężynie z tłumieniem

**Polecenie dla ChatGPT:**

> Podaj kod Pythona realizujący ruch ciała na sprężynie z tłumieniem.


```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.integrate

# Parametry ruchu harmonicznego tłumionego
k = 50  # stała sprężystości (N/m)
c = 1.5  # współczynnik tłumienia (kg/s)
m = 2.0  # masa (kg)
x0 = 1.0  # wychylenie początkowe (m)
v0 = 0.0  # prędkość początkowa (m/s)

# Funkcja opisująca układ równań różniczkowych
def damped_spring(t, y):
    x, v = y
    dxdt = v
    dvdt = -(k/m) * x - (c/m) * v
    return [dxdt, dvdt]

# Warunki początkowe: x, v
initial_conditions = [x0, v0]

# Czas symulacji
t = np.linspace(0, 10, 500)

# Rozwiązanie równań
solution = scipy.integrate.solve_ivp(
    damped_spring, [t[0], t[-1]], initial_conditions, t_eval=t
)

# Położenie w czasie
x = solution.y[0]

# Wykres położenia od czasu
plt.figure(figsize=(10, 6))
plt.plot(t, x, label="Położenie sprężyny z tłumieniem", linewidth=2)
plt.title("Położenie od czasu dla sprężyny z tłumieniem")
plt.xlabel("Czas (s)")
plt.ylabel("Położenie (m)")
plt.axhline(0, color="black", linewidth=0.8, linestyle="--")
plt.grid(True)
plt.legend()
plt.show()
```

# Podstawowe zasady dynamiki Newtona

Zasady dynamiki sformułowane przez Isaaca Newtona stanowią fundament mechaniki klasycznej. Opisują one relacje między ruchem ciał a siłami na nie działającymi w inercjalnych układach odniesienia.

---

### I zasada dynamiki (Zasada bezwładności)

Jeżeli na ciało nie działa żadna siła zewnętrzna lub siły działające równoważą się (siła wypadkowa jest równa zeru), to ciało pozostaje w spoczynku lub porusza się ruchem jednostajnym prostoliniowym.

W ujęciu matematycznym:


$$\vec{F}_{\text{wyp}} = \sum \vec{F} = 0 \implies \vec{v} = \text{const}$$

*Gdy $\vec{v} = 0$, ciało spoczywa; gdy $\vec{v} \neq 0$, porusza się ze stałą prędkością po linii prostej.*

---

### II zasada dynamiki (Zasada przyspieszenia)

Jeżeli na ciało działa niezrównoważona siła wypadkowa, to ciało porusza się z przyspieszeniem wprost proporcjonalnym do wartości tej siły i odwrotnie proporcjonalnym do masy ciała. Przyspieszenie ma ten sam kierunek i zwrot co siła wypadkowa.

W ujęciu wektorowym:


$$\vec{a} = \frac{\vec{F}_{\text{wyp}}}{m} \quad \iff \quad \vec{F}_{\text{wyp}} = m \cdot \vec{a}$$

W uogólnionej postaci relatywistycznej oraz dla ciał o zmiennej masie definiuje się ją jako pochodną pędu $\vec{p} = m \vec{v}$ po czasie:


$$\vec{F}_{\text{wyp}} = \frac{d\vec{p}}{dt}$$

---

### III zasada dynamiki (Zasada akcji i reakcji)

Oddziaływania ciał są zawsze wzajemne. Jeżeli ciało $A$ działa na ciało $B$ siłą $\vec{F}_{A \to B}$ (akcja), to ciało $B$ działa na ciało $A$ siłą $\vec{F}_{B \to A}$ (reakcja) o tej samej wartości i kierunku, lecz o przeciwnym zwrocie.

W ujęciu matematycznym:


$$\vec{F}_{A \to B} = -\vec{F}_{B \to A}$$

*Ważna uwaga:* Siły te nie równoważą się, ponieważ są przyłożone do dwóch różnych ciał.

---

## Podsumowanie

| Prawo | Zasada | Przykład |
| --- | --- | --- |
| **I zasada dynamiki** | Gdy $\vec{F}_{\text{wyp}} = 0$, ciało spoczywa lub porusza się ruchem jednostajnym prostoliniowym ($\vec{v} = \text{const}$). | Pasażer gwałtownie hamującego autobusu leci do przodu na skutek bezwładności własnego ciała. |
| **II zasada dynamiki** | Siła wypadkowa nadaje ciału przyspieszenie proporcjonalne do siły i odwrotnie proporcjonalne do masy: $\vec{F} = m \vec{a}$. | Pchanie pustego wózka sklepowego wymaga mniejszej siły niż pchanie wózka pełnego, aby nadać im takie samo przyspieszenie. |
| **III zasada dynamiki** | Oddziaływania są wzajemne: każdej akcji towarzyszy równa co do wartości i przeciwnie skierowana reakcja ($\vec{F}_{A \to B} = -\vec{F}_{B \to A}$). | Odrzut wystrzeliwanego pocisku z karabinu lub ruch rakiety napędzanej gazami wylotowymi. |

### Rzut ukośny dla różnych kątów


```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
v0 = 50       # Initial velocity (m/s)
angles = [15, 45, 60]  # Launch angles (degrees)
g = 9.81      # Gravity (m/s²)

# Create figure
plt.figure(figsize=(10,6))

# Calculate and plot for each angle
for theta_deg in angles:
    # Convert angle to radians
    theta = np.radians(theta_deg)

    # Initial velocity components
    v0x = v0 * np.cos(theta)
    v0y = v0 * np.sin(theta)

    # Time calculations
    t_flight = (2 * v0y) / g
    t = np.linspace(0, t_flight, 1000)

    # Position calculations
    x = v0x * t
    y = v0y * t - 0.5 * g * t**2

    # Plot trajectory
    plt.plot(x, y, label=f'θ = {theta_deg}°')

    # Calculate key quantities
    H = (v0y**2) / (2*g)
    R = v0x * t_flight
    print(f"\nAngle: {theta_deg}°")
    print(f"Time of Flight: {t_flight:.2f} s")
    print(f"Maximum Height: {H:.2f} m")
    print(f"Horizontal Range: {R:.2f} m")

# Customize plot
plt.title('Projectile Motion for Different Launch Angles')
plt.xlabel('Horizontal Distance (m)')
plt.ylabel('Vertical Height (m)')
plt.ylim(-5, 110)  # Jawne ustawienie limitów osi Y od -5 do 110 m
plt.legend()
plt.grid(True)
plt.show()
```

### Animacja rzutu ukośnego z oporem i bez oporu powietrza


```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# Parametry
m = 1.0  # Masa [kg]
g = 9.81  # Przyspieszenie grawitacyjne [m/s^2]
k = 0.1  # Współczynnik oporu powietrza [kg/s]
v0 = 20  # Prędkość początkowa [m/s]
theta = np.radians(45)  # Kąt początkowy [stopnie -> radiany]
dt = 0.02  # Krok czasowy [s]
T = 5  # Czas symulacji [s]
elasticity = 0.7  # Współczynnik sprężystości odbicia

# Funkcja obliczająca trajektorię bez oporu powietrza
def trajectory_no_drag(v0, theta, g, dt, T, elasticity):
    x, y = [0], [0]
    vx, vy = v0 * np.cos(theta), v0 * np.sin(theta)
    t = 0
    while t < T:
        x.append(x[-1] + vx * dt)
        y_new = y[-1] + vy * dt - 0.5 * g * dt**2
        vy -= g * dt
        if y_new < 0:  # Odbicie od ziemi
            vy = -vy * elasticity
            y_new = 0
        y.append(y_new)
        t += dt
    return x, y

# Funkcja obliczająca trajektorię z oporem powietrza
def trajectory_with_drag(v0, theta, g, k, m, dt, T, elasticity):
    vx, vy = v0 * np.cos(theta), v0 * np.sin(theta)
    x, y = [0], [0]
    t = 0
    while t < T:
        ax = - (k / m) * vx
        ay = -g - (k / m) * vy
        vx += ax * dt
        vy += ay * dt
        x.append(x[-1] + vx * dt)
        y_new = y[-1] + vy * dt
        if y_new < 0:  # Odbicie od ziemi
            vy = -vy * elasticity
            y_new = 0
        y.append(y_new)
        t += dt
    return x, y

# Oblicz trajektorie
x_no_drag, y_no_drag = trajectory_no_drag(v0, theta, g, dt, T, elasticity)
x_drag, y_drag = trajectory_with_drag(v0, theta, g, k, m, dt, T, elasticity)

# Inicjalizacja animacji
fig, ax = plt.subplots(figsize=(8, 5))
ax.set_xlim(0, max(max(x_no_drag), max(x_drag)))
ax.set_ylim(0, max(max(y_no_drag), max(y_drag)))
ax.set_xlabel("Odległość [m]")
ax.set_ylabel("Wysokość [m]")
ax.set_title("Rzut ukośny z i bez oporu powietrza (z odbiciami)")
ax.grid()

point_no_drag, = ax.plot([], [], 'bo', label="Bez oporu")
point_drag, = ax.plot([], [], 'ro', label="Z oporem")
traj_no_drag, = ax.plot([], [], 'b--', alpha=0.5)
traj_drag, = ax.plot([], [], 'r--', alpha=0.5)
ax.legend()

# Funkcja inicjalizacyjna
def init():
    point_no_drag.set_data([], [])
    point_drag.set_data([], [])
    traj_no_drag.set_data([], [])
    traj_drag.set_data([], [])
    return point_no_drag, point_drag, traj_no_drag, traj_drag

# Funkcja aktualizująca animację
def update(frame):
    if frame < len(x_no_drag):
        point_no_drag.set_data([x_no_drag[frame]], [y_no_drag[frame]])
        traj_no_drag.set_data(x_no_drag[:frame+1], y_no_drag[:frame+1])

    if frame < len(x_drag):
        point_drag.set_data([x_drag[frame]], [y_drag[frame]])
        traj_drag.set_data(x_drag[:frame+1], y_drag[:frame+1])

    return point_no_drag, point_drag, traj_no_drag, traj_drag

# Tworzenie animacji
frames = max(len(x_no_drag), len(x_drag))
ani = FuncAnimation(fig, update, frames=frames, init_func=init, blit=False, interval=20)

# Wyświetlenie animacji w Colabie
plt.close()
HTML(ani.to_html5_video())
```

### Transfer ładunku z Ziemi na Marsa


```python
# prompt: Give me file.gif

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# Constants (in AU and days)
R_earth = 1.0         # Earth's orbit radius
R_mars = 1.524        # Mars' orbit radius
T_earth = 365.25      # Earth orbital period (days)
T_mars = 687.0        # Mars orbital period (days)
T_transfer = 259.0    # Hohmann transfer time (days)

# Hohmann transfer ellipse parameters
a = (R_earth + R_mars) / 2.0  # Semi-major axis
e = (R_mars - R_earth) / (R_mars + R_earth)  # Eccentricity

# Correct Mars' initial position so that it is 44 degrees ahead of Earth at launch
theta_mars_initial = np.radians(44)  # Convert 44 degrees to radians

def get_planet_position(radius, period, time, initial_theta=0):
    theta = (2 * np.pi * time) / period + initial_theta
    x = radius * np.cos(theta)
    y = radius * np.sin(theta)
    return x, y

# Time settings
num_frames = 200  # Number of animation frames
time_days = np.linspace(0, T_transfer, num_frames)  # Simulated time

# Planet positions over time
x_earth, y_earth = get_planet_position(R_earth, T_earth, time_days)
x_mars, y_mars = get_planet_position(R_mars, T_mars, time_days, theta_mars_initial)

# Transfer orbit calculations
theta_transfer = np.linspace(0, np.pi, num_frames)
r_transfer = a * (1 - e**2) / (1 + e * np.cos(theta_transfer))
x_transfer = r_transfer * np.cos(theta_transfer)
y_transfer = r_transfer * np.sin(theta_transfer)

# Full orbits for static reference
theta_full = np.linspace(0, 2 * np.pi, 500)
x_earth_orbit = R_earth * np.cos(theta_full)
y_earth_orbit = R_earth * np.sin(theta_full)

x_mars_orbit = R_mars * np.cos(theta_full)
y_mars_orbit = R_mars * np.sin(theta_full)

# Create figure
fig, ax = plt.subplots(figsize=(8, 8))
ax.plot(x_earth_orbit, y_earth_orbit, 'b--', label="Earth's Orbit")
ax.plot(x_mars_orbit, y_mars_orbit, 'r--', label="Mars' Orbit")
ax.plot(0, 0, 'yo', markersize=12, label='Sun')
ax.plot(x_transfer, y_transfer, 'g--', alpha=0.5, label='Transfer Ellipse')

ax.set_xlim(-R_mars - 0.5, R_mars + 0.5)
ax.set_ylim(-R_mars - 0.5, R_mars + 0.5)
ax.set_aspect('equal')
ax.set_xlabel('x (AU)')
ax.set_ylabel('y (AU)')
ax.set_title('Hohmann Transfer: Earth to Mars')
ax.legend()
ax.grid(True)

# Initialize animation objects
spacecraft, = ax.plot([], [], 'ko', markersize=8, label='Spacecraft')
earth_marker, = ax.plot([], [], 'bo', markersize=10, label='Earth')
mars_marker, = ax.plot([], [], 'ro', markersize=10, label='Mars')
anomaly_text = ax.text(0.05, 0.95, '', transform=ax.transAxes)

def init():
    spacecraft.set_data([], [])
    earth_marker.set_data([], [])
    mars_marker.set_data([], [])
    anomaly_text.set_text('')
    return spacecraft, earth_marker, mars_marker, anomaly_text

# Animation function
def animate(i):
    spacecraft.set_data([x_transfer[i]], [y_transfer[i]])
    earth_marker.set_data([x_earth[i]], [y_earth[i]])
    mars_marker.set_data([x_mars[i]], [y_mars[i]])
    anomaly_text.set_text(f"Day: {int(time_days[i])}")
    return spacecraft, earth_marker, mars_marker, anomaly_text

# Create animation
ani = FuncAnimation(fig, animate, frames=num_frames, init_func=init, interval=50, blit=True)
plt.close()
HTML(ani.to_html5_video())
```
