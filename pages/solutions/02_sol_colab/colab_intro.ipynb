```python
# Sumowanie ciągu geometrycznego
# Ciąg geometryczny to ciąg liczb, w którym każdy kolejny wyraz powstaje przez pomnożenie poprzedniego przez stałą wartość q

# Definicja zmiennych
a1 = 2
q = 3
n = 5

# Obliczenie sumy n wyrazów ciągu geometrycznego
if q == 1:
    S_n = a1 * n
else:
    S_n = a1 * (1 - q**n) / (1 - q)

print(f"Suma pierwszych {n} wyrazów ciągu geometrycznego (dla a1={a1}, q={q}) wynosi: {S_n}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import Video

# Parametry wahadła
g = 9.81
L = 1.0
theta0 = np.pi / 4  # kąt początkowy

# Czas
t = np.linspace(0, 10, 300)

# Ruch wahadła (małe wychylenia)
theta = theta0 * np.cos(np.sqrt(g / L) * t)

# Współrzędne kulki
x = L * np.sin(theta)
y = -L * np.cos(theta)

# Tworzenie wykresu
fig, ax = plt.subplots(figsize=(5, 5))
ax.set_xlim(-1.2, 1.2)
ax.set_ylim(-1.2, 0.2)
ax.set_aspect('equal')
ax.grid()

line, = ax.plot([], [], 'o-', lw=2)

def update(frame):
    line.set_data([0, x[frame]], [0, y[frame]])
    return line,

ani = FuncAnimation(fig, update, frames=len(t), interval=30)

# Zapis do pliku MP4
ani.save("video.mp4", writer="ffmpeg", fps=30)

plt.close()

print("Zapisano plik video.mp4")

# Osadzenie filmu w notebooku
Video("video.mp4", embed=True)
```
